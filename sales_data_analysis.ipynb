```python
import pandas as pd
import random
from datetime import datetime, timedelta

# -----------------------------
# SETTINGS
# -----------------------------

random.seed(42)

num_orders = 1000

first_date = datetime(2026, 1, 1)
last_date = datetime(2026, 12, 31)

# -----------------------------
# DATA OPTIONS
# -----------------------------

first_names = [
    "Aarav", "Priya", "Rahul", "Ananya", "Vikram",
    "Sneha", "Arjun", "Kavya", "Rohan", "Neha",
    "Aditya", "Pooja", "Karan", "Ishita", "Manish",
    "Divya", "Nikhil", "Meera", "Sanjay", "Aisha"
]

last_names = [
    "Sharma", "Singh", "Kumar", "Patel", "Reddy",
    "Verma", "Gupta", "Iyer", "Nair", "Joshi",
    "Mehta", "Rao", "Das", "Malhotra", "Chopra"
]

products = {
    "Electronics": {
        "Laptop": 60000,
        "Smartphone": 30000,
        "Monitor": 15000,
        "Keyboard": 2000,
        "Mouse": 1000,
        "Headphones": 2500,
        "Tablet": 25000
    },

    "Furniture": {
        "Office Chair": 5000,
        "Desk": 10000,
        "Bookshelf": 7000,
        "Sofa": 25000,
        "Table": 8000,
        "Cabinet": 6000
    },

    "Clothing": {
        "T-Shirt": 800,
        "Jeans": 1800,
        "Jacket": 3000,
        "Shoes": 2500,
        "Shirt": 1500,
        "Dress": 2200
    },

    "Office Supplies": {
        "Notebook": 150,
        "Pen Set": 200,
        "Printer Paper": 400,
        "Stapler": 250,
        "Calculator": 600,
        "File Folder": 100
    }
}

regions = ["North", "South", "East", "West"]

payment_methods = [
    "UPI",
    "Credit Card",
    "Debit Card",
    "Cash",
    "Net Banking"
]

# Different profit margins by category
profit_margins = {
    "Electronics": (0.08, 0.18),
    "Furniture": (0.12, 0.25),
    "Clothing": (0.18, 0.35),
    "Office Supplies": (0.10, 0.22)
}

# -----------------------------
# GENERATE DATA
# -----------------------------

data = []

for i in range(num_orders):

    # Order ID
    order_id = 1001 + i

    # Random date
    days_difference = (last_date - first_date).days
    random_days = random.randint(0, days_difference)
    order_date = first_date + timedelta(days=random_days)

    # Customer
    customer_name = (
        random.choice(first_names)
        + " "
        + random.choice(last_names)
    )

    # Category
    category = random.choice(list(products.keys()))

    # Product
    product = random.choice(list(products[category].keys()))

    # Unit price
    unit_price = products[category][product]

    # Quantity
    quantity = random.randint(1, 10)

    # Sales
    sales = quantity * unit_price

    # Profit
    min_margin, max_margin = profit_margins[category]
    profit_margin = random.uniform(min_margin, max_margin)

    profit = round(sales * profit_margin, 2)

    # Region
    region = random.choice(regions)

    # Payment method
    payment_method = random.choice(payment_methods)

    # Add row
    data.append([
        order_id,
        order_date.strftime("%Y-%m-%d"),
        customer_name,
        product,
        category,
        region,
        quantity,
        unit_price,
        sales,
        profit,
        payment_method
    ])

# -----------------------------
# CREATE DATAFRAME
# -----------------------------

columns = [
    "Order_ID",
    "Order_Date",
    "Customer_Name",
    "Product",
    "Category",
    "Region",
    "Quantity",
    "Unit_Price",
    "Sales",
    "Profit",
    "Payment_Method"
]

df = pd.DataFrame(data, columns=columns)

# -----------------------------
# SAVE CSV
# -----------------------------

df.to_csv("sales_dataset_1000.csv", index=False)

print("Dataset created successfully!")
print("Rows:", len(df))
print("File: sales_dataset_1000.csv")

# Show first 10 rows
print("\nFirst 10 rows:")
print(df.head(10))

# Basic summary
print("\nTotal Sales:", df["Sales"].sum())
print("Total Profit:", round(df["Profit"].sum(), 2))
print("Total Quantity:", df["Quantity"].sum())
```
